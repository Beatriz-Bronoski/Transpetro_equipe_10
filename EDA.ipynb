```python
# --- DATA PREPARATION & CLEANING ---

# 1. Date Parsing & Standardization
# We need to handle mixed formats (e.g., DD/MM/YYYY and MM/DD/YYYY) carefully.
def parse_dates_smart(date_series):
    # Try parsing with dayfirst=True (common in Brazil)
    return pd.to_datetime(date_series, dayfirst=True, errors='coerce')

relatorio_iws['Data_Clean'] = parse_dates_smart(relatorio_iws['Data'])
resultado_eventos['startGMTDate'] = pd.to_datetime(resultado_eventos['startGMTDate'], errors='coerce')

print("Date parsing complete. Null dates in inspections:", relatorio_iws['Data_Clean'].isnull().sum())
```

```python
# 2. Target Variable Engineering: Fouling Score (0-100)
import re

def extract_fouling_score(text):
    if pd.isna(text):
        return None
    text = str(text).lower()
    
    # Case 1: Direct percentage (e.g., '100%', '10%')
    match = re.search(r'(\d+)%', text)
    if match:
        return float(match.group(1))
    
    # Case 2: Norman Scale (1-5 often used, mapping to 0-100)
    # Assuming 1=Clean(0), 5=Heavy(100) roughly, or just keeping raw if it's a scale.
    # Let's look for single digits if no % exists
    match_digit = re.search(r'^\s*(\d+)\s*$', text)
    if match_digit:
        val = float(match_digit.group(1))
        if val <= 5: # Likely Norman scale
            return val * 20 # Map 1-5 to 20-100 roughly
        return val
        
    # Case 3: Text Descriptions (Heuristic Mapping)
    if 'limo' in text and 'craca' not in text:
        return 20 # Light fouling
    if 'craca' in text:
        return 80 # Heavy fouling
    if 'alga' in text:
        return 40 # Medium fouling
        
    return None

# Apply to key columns
relatorio_iws['Fouling_Score_General'] = relatorio_iws['Condição geral da embarcação'].apply(extract_fouling_score)
relatorio_iws['Fouling_Score_Hull'] = relatorio_iws['Condição do fundo chato'].apply(extract_fouling_score)

# Fill missing scores with an average or default if needed, for now let's keep NaNs
print(relatorio_iws[['Condição geral da embarcação', 'Fouling_Score_General']].head(10))
```

```python
# 3. Advanced Merging: Propagating Inspection Data
# We want to associate each voyage event with the *most recent* inspection BEFORE that voyage.

# Sort both datasets by date
relatorio_iws = relatorio_iws.dropna(subset=['Data_Clean'])
relatorio_iws = relatorio_iws.sort_values('Data_Clean')
resultado_eventos = resultado_eventos.sort_values('startGMTDate')

# Ensure ship names match exactly (uppercase already done)
relatorio_iws['Embarcação'] = relatorio_iws['Embarcação'].str.strip().str.upper()
resultado_eventos['shipName'] = resultado_eventos['shipName'].str.strip().str.upper()

# Use merge_asof to find the nearest prior inspection
# Note: merge_asof requires sorted keys
merged_data = pd.merge_asof(
    resultado_eventos,
    relatorio_iws[['Embarcação', 'Data_Clean', 'Fouling_Score_General', 'Fouling_Score_Hull']],
    left_on='startGMTDate',
    right_on='Data_Clean',
    left_by='shipName',
    right_by='Embarcação',
    direction='backward' # Look for the closest past date
)

# Calculate 'Days Since Inspection'
merged_data['Days_Since_Inspection'] = (merged_data['startGMTDate'] - merged_data['Data_Clean']).dt.days

print("Merge complete. Sample of new columns:")
print(merged_data[['shipName', 'startGMTDate', 'Data_Clean', 'Fouling_Score_General', 'Days_Since_Inspection']].head())
```

```python
# 4. Final Dataset for Model
# Now you have 'merged_data' with:
# - Operational data (Speed, Consumption, Weather)
# - Fouling State (Fouling_Score_General)
# - Time factor (Days_Since_Inspection)

# You can now drop rows where we have no inspection data if strict training is needed
final_df = merged_data.dropna(subset=['Fouling_Score_General'])
print(f"Final dataset shape: {final_df.shape}")
```

```python
# 5. Incorporating AIS Data
import glob

# Path to AIS data
ais_path = 'Dados AIS frota TP/*.csv'
ais_files = glob.glob(ais_path)

print(f"Found {len(ais_files)} AIS files.")

# Load and concatenate all AIS files
ais_dfs = []
for f in ais_files:
    try:
        df = pd.read_csv(f)
        # Extract ship name from filename if not in columns, but it seems to be in 'NOME'
        ais_dfs.append(df)
    except Exception as e:
        print(f"Error reading {f}: {e}")

if ais_dfs:
    ais_df = pd.concat(ais_dfs, ignore_index=True)
    print("AIS Data Loaded. Shape:", ais_df.shape)
    print(ais_df.head())
else:
    print("No AIS data loaded.")
```

```python
# Preprocessing AIS Data
if 'ais_df' in locals():
    # Convert timestamp
    ais_df['DATAHORA'] = pd.to_datetime(ais_df['DATAHORA'], errors='coerce')
    
    # Create Date column for daily aggregation
    ais_df['Date'] = ais_df['DATAHORA'].dt.floor('D')
    
    # Clean Speed (remove outliers/errors)
    # Assuming speed is in knots. > 30 knots is unlikely for these tankers.
    ais_df = ais_df[(ais_df['VELOCIDADE'] >= 0) & (ais_df['VELOCIDADE'] < 30)]
    
    # Normalize Ship Name
    ais_df['NOME'] = ais_df['NOME'].str.strip().str.upper()
    
    print("AIS Data Preprocessed.")
```

```python
# Aggregating AIS Data by Day and Ship
if 'ais_df' in locals():
    ais_daily = ais_df.groupby(['NOME', 'Date']).agg(
        AIS_Speed_Mean=('VELOCIDADE', 'mean'),
        AIS_Speed_Std=('VELOCIDADE', 'std'),
        AIS_Data_Count=('VELOCIDADE', 'count')
    ).reset_index()
    
    print("AIS Daily Aggregation Complete. Shape:", ais_daily.shape)
    print(ais_daily.head())
```

```python
# Merging AIS Data with Main Dataset
if 'merged_data' in locals() and 'ais_daily' in locals():
    # Create Date column in merged_data for joining
    merged_data['Date'] = merged_data['startGMTDate'].dt.floor('D')
    
    # Merge
    # We use left join to keep all events, adding AIS info where available
    merged_data_ais = pd.merge(
        merged_data,
        ais_daily,
        left_on=['shipName', 'Date'],
        right_on=['NOME', 'Date'],
        how='left'
    )
    
    print("Merge with AIS complete. New Shape:", merged_data_ais.shape)
    print(merged_data_ais[['shipName', 'Date', 'AIS_Speed_Mean', 'Fouling_Score_General']].head())
    
    # Check correlation
    corr = merged_data_ais[['Fouling_Score_General', 'AIS_Speed_Mean']].corr()
    print("\nCorrelation between Fouling Score and AIS Speed:")
    print(corr)
```
